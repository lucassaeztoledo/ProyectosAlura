### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda1 = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda1.head()
```

#1. Análisis de facturación



```python
ingreso_tienda1 = tienda1['Precio'].sum()
ingreso_tienda2 = tienda2['Precio'].sum()
ingreso_tienda3 = tienda3['Precio'].sum()
ingreso_tienda4 = tienda4['Precio'].sum()

print(f"Ingreso total Tienda 1: {ingreso_tienda1:.2f}")
print(f"Ingreso total Tienda 2: {ingreso_tienda2:.2f}")
print(f"Ingreso total Tienda 3: {ingreso_tienda3:.2f}")
print(f"Ingreso total Tienda 4: {ingreso_tienda4:.2f}")
```

# 2. Ventas por categoría

# 3. Calificación promedio de la tienda


```python
calificacion_promedio_tienda1 = tienda1['Calificación'].mean()
calificacion_promedio_tienda2 = tienda2['Calificación'].mean()
calificacion_promedio_tienda3 = tienda3['Calificación'].mean()
calificacion_promedio_tienda4 = tienda4['Calificación'].mean()

print(f"Calificación promedio Tienda 1: {calificacion_promedio_tienda1:.2f}")
print(f"Calificación promedio Tienda 2: {calificacion_promedio_tienda2:.2f}")
print(f"Calificación promedio Tienda 3: {calificacion_promedio_tienda3:.2f}")
print(f"Calificación promedio Tienda 4: {calificacion_promedio_tienda4:.2f}")
```

# 4. Productos más y menos vendidos

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Función para obtener los productos más y menos vendidos
def obtener_productos_top_bottom(df, nombre_tienda, top_n=5):
    conteo_productos = df['Producto'].value_counts()
    productos_mas_vendidos = conteo_productos.head(top_n)
    productos_menos_vendidos = conteo_productos.tail(top_n)

    print(f"\n--- {nombre_tienda} ---")
    print(f"Productos más vendidos (Top {top_n}):\n{productos_mas_vendidos}")
    print(f"Productos menos vendidos (Bottom {top_n}):\n{productos_menos_vendidos}")

    # Visualización
    fig, axes = plt.subplots(1, 2, figsize=(18, 6))
    fig.suptitle(f'Productos más y menos vendidos en {nombre_tienda}', fontsize=16)

    sns.barplot(x=productos_mas_vendidos.index, y=productos_mas_vendidos.values, ax=axes[0], palette='viridis', hue=productos_mas_vendidos.index, legend=False)
    axes[0].set_title(f'Top {top_n} Productos más vendidos')
    axes[0].set_xlabel('Producto')
    axes[0].set_ylabel('Cantidad de Ventas')
    axes[0].tick_params(axis='x', rotation=45)

    sns.barplot(x=productos_menos_vendidos.index, y=productos_menos_vendidos.values, ax=axes[1], palette='magma', hue=productos_menos_vendidos.index, legend=False)
    axes[1].set_title(f'Top {top_n} Productos menos vendidos')
    axes[1].set_xlabel('Producto')
    axes[1].set_ylabel('Cantidad de ventas')
    axes[1].tick_params(axis='x', rotation=45)

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

obtener_productos_top_bottom(tienda1, 'Tienda 1')
obtener_productos_top_bottom(tienda2, 'Tienda 2')
obtener_productos_top_bottom(tienda3, 'Tienda 3')
obtener_productos_top_bottom(tienda4, 'Tienda 4')
```

# 5. Envío promedio por tienda

```python
costo_envio_promedio_tienda1 = tienda1['Costo de envío'].mean()
costo_envio_promedio_tienda2 = tienda2['Costo de envío'].mean()
costo_envio_promedio_tienda3 = tienda3['Costo de envío'].mean()
costo_envio_promedio_tienda4 = tienda4['Costo de envío'].mean()

print(f"Costo de envío promedio Tienda 1: {costo_envio_promedio_tienda1:.2f}")
print(f"Costo de envío promedio Tienda 2: {costo_envio_promedio_tienda2:.2f}")
print(f"Costo de envío promedio Tienda 3: {costo_envio_promedio_tienda3:.2f}")
print(f"Costo de envío promedio Tienda 4: {costo_envio_promedio_tienda4:.2f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# --- 1. Gráfico de barras para el Ingreso Total por Tienda ---
ingresos_totales = pd.DataFrame({
    'Tienda': ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'],
    'Ingreso total': [ingreso_tienda1, ingreso_tienda2, ingreso_tienda3, ingreso_tienda4]
})

plt.figure(figsize=(10, 6))
sns.barplot(x='Tienda', y='Ingreso total', data=ingresos_totales, palette='viridis', hue='Tienda', legend=False)
plt.title('Ingreso total por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Ingreso total')
plt.ticklabel_format(style='plain', axis='y')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# --- 2. Gráfico de torta para Ventas por Categoría de Producto (acumulado de todas las tiendas) ---
# Unir todas las series de ventas por categoría
ventas_por_categoria_consolidado = pd.concat([
    ventas_por_categoria_tienda1,
    ventas_por_categoria_tienda2,
    ventas_por_categoria_tienda3,
    ventas_por_categoria_tienda4
]).groupby(level=0).sum().sort_values(ascending=False)

plt.figure(figsize=(10, 8))
plt.pie(ventas_por_categoria_consolidado, labels=ventas_por_categoria_consolidado.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Distribución de ventas por Categoría de Producto (todas las Tiendas)')
plt.axis('equal') # Para que el círculo sea perfecto
plt.show()

# --- 3. Gráfico de Área para Ventas Mensuales Totales (Todas las Tiendas) ---
# Concatenar todos los dataframes de tiendas
df_all_stores = pd.concat([tienda1, tienda2, tienda3, tienda4])

# Convertir 'Fecha de Compra' a formato de fecha y manejar errores
df_all_stores['Fecha de Compra'] = pd.to_datetime(df_all_stores['Fecha de Compra'], format='%d/%m/%Y', errors='coerce')

# Eliminar filas con fechas no válidas (NaN) si las hay
df_all_stores.dropna(subset=['Fecha de Compra'], inplace=True)

# Agrupar por mes y año, luego sumar los precios
ventas_mensuales = df_all_stores.groupby(df_all_stores['Fecha de Compra'].dt.to_period('M'))['Precio'].sum().reset_index()
ventas_mensuales['Fecha de Compra'] = ventas_mensuales['Fecha de Compra'].dt.to_timestamp() # Convertir de Period a Timestamp para el gráfico

plt.figure(figsize=(12, 7))
plt.fill_between(ventas_mensuales['Fecha de Compra'], ventas_mensuales['Precio'], color="skyblue", alpha=0.4)
plt.plot(ventas_mensuales['Fecha de Compra'], ventas_mensuales['Precio'], color="Slateblue", alpha=0.6)
plt.title('Ventas mensuales totales (Todas las Tiendas)')
plt.xlabel('Fecha')
plt.ylabel('Ingreso total')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
